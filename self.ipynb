preprocess(path：dir, SPH：s, SOP：s, interictal_thr, window_length：s, window_overlap：s, elec_names) -> windows,meta

````python
def preprocess(path, SPH, SOP, interictal_thr, window_length, window_overlap, elec_names):
    pid = os.path.basename(os.path.normpath(path))
    recs = read_summary(_summary(path))
    seize = sorted(s for r in recs for s in r["seizures"]) # 从dataframe取出发作[s,e]

    W = int(round(window_length * FS_OUT))
    O = int(round(window_overlap * FS_OUT))
    rows = {c: [] for c in elec_names}
    wins = {c: [] for c in elec_names}
    # skipped：存储该跳过的文件
    # short：存储长度不够的窗口
    skipped, short = [], []

    # 读取各edf文件数据
    for rec in recs:
        x, rate = read_edf(os.path.join(path, rec["file"]), elec_names)
        if x is None: #若为空则跳过
            skipped.append(rec["file"])
            continue
        # # 滤波
        # x = bandpass_filter(resample_signal(x, rate, FS_OUT), FS_OUT, *BAND, ORDER)
        # 切窗
        win, starts = make_windows(x, W, O)
        # 跳过没有切窗成功的文件
        if not len(starts):
            short.append(rec["file"])
            continue

        rest = x.shape[-1] - (int(starts[-1]) + W) # 计算切窗剩余长度
        if rest > 0:
            short.append("%s(丢 %.1fs)" % (rec["file"], rest / FS_OUT))

        # 打标签
        labs = [label_window(rec["t0"] + s / FS_OUT, rec["t0"] + (s + W) / FS_OUT,
                             seize, SPH, SOP, interictal_thr) for s in starts]
        # 分别写出各通道metadata
        for i, c in enumerate(elec_names):
            n0 = len(rows[c])
            for k, s in enumerate(starts):
                d = labs[k]
                rows[c].append(dict(
                    edf_file=rec["file"], patient_id=pid, record_id=rec["file"].replace(".edf", ""),
                    channel=c, seizure_id=int(d["seizure_id"]), window_id=n0 + k,
                    start_abs_s=round(rec["t0"] + s / FS_OUT, 3),
                    end_abs_s=round(rec["t0"] + (s + W) / FS_OUT, 3),
                    center_abs_s=round(rec["t0"] + (s + W / 2.0) / FS_OUT, 3),
                    raw_start_sample=int(round(s * FS_IN / FS_OUT)),
                    raw_end_sample=int(round((s + W) * FS_IN / FS_OUT)),
                    start_sample=int(s), end_sample=int(s + W),
                    prev_seizure_id=int(d["prev_seizure_id"]), next_seizure_id=int(d["next_seizure_id"]),
                    time_from_previous_seizure_start_s=round(d["time_from_previous_seizure_start_s"], 3),
                    time_from_previous_seizure_end_s=round(d["time_from_previous_seizure_end_s"], 3),
                    time_to_next_seizure_start_s=round(d["time_to_next_seizure_start_s"], 3),
                    time_to_next_seizure_end_s=round(d["time_to_next_seizure_end_s"], 3),
                    label=int(d["label"]), usable=int(d["usable"]),
                    excluded_reason=d["excluded_reason"]))
            wins[c].append(win[i])

    if skipped:
        print("[警告] %s 跳过 %d 份 EDF: %s" % (pid, len(skipped), skipped))
    if short:
        print("[边界] %s 末尾不足整窗: %s" % (pid, short))

    windows, meta = {}, {}
    for c in elec_names:
        # 把一个通道的所有窗口按dim=0（窗口）拼接成一个完整窗口数组
        windows[c] = np.concatenate(wins[c]) if wins[c] else np.zeros((0, W), np.float32)
        # 把所有元数据拼接成一个metadata
        meta[c] = pd.DataFrame(rows[c], columns=COLS)

    # 打印结果
    _print_counts(meta[elec_names[0]], "[%s]" % pid)
    for c in elec_names:
        print("[%s] %s 窗 %d" % (pid, c, len(meta[c])))

    return windows, meta
````

# encoder

x.shape = (batch=32,time=10s*256,elec=1,channal=1)

1.卷积 -> 归一 -> 激活 -> 时间降采样，降低时间分辨率并保留局部强响应

2.只对确定的6个电极的数据自编码;去噪，提取关键模式、特征

3.加入输入扰动，通过去噪重构迫使 Encoder 学习稳定、具有鲁棒性的表征

4.深度可分离卷积，降低参数

````python
import numpy as np

def build_encoder(input_shape,filters=filters):
    inputs = L.Input(shape=input_shape)
    x = inputs

    for i in range(3):
        # 深度可分离卷积
        x = L.SeparableConv1D(
                     filters=filters, # 第一次卷积时，channal：1 -> 128
                     kernel_size=3,
                     strides=1,
                     padding='same',
                     use_bias=True
        )(x)

        x = L.BatchNormalization()(x)
        x = L.ReLU()(x)
        x = L.MaxPool1D(pool_size=2,strides=2,padding='same')(x) # time / 2

    outputs = x

    return tf.keras.Model(inputs=inputs,outputs=outputs)

encoder = build_encoder(input_shape=(2000,1))
encoder.summary()
````

````python
def build_decoder(input_shape,filters=filters):

    inputs = L.Input(shape=input_shape)
    x = inputs

    for i in range(2):
        x = L.Conv1DTranspose(
            filters=filters,
            kernel_size=3,
            strides=2,
            padding='same',
            use_bias=True
        )(x)
        x = L.BatchNormalization()(x)
        x = L.ReLU()(x)

    # channal：128 -> 1
    x = L.Conv1DTranspose(filters=1,kernel_size=3,strides=2,padding='same',use_bias=True)(x)

    # 最后一层不加激活，避免重构误差
    # x = L.BatchNormalization()(x)
    # x = L.ReLU()(x)

    outputs = x

    return tf.keras.Model(inputs=inputs,outputs=outputs)

decoder = build_decoder(input_shape=encoder.output_shape[1:])
decoder.summary()
````

自编码器

````python
def build_autoencoder(input_shape,filters=filters):
    # 输入占位
    inputs = L.Input(shape=input_shape)

    # 初始化encoder、decoder
    encoder = build_encoder(input_shape,filters)
    encoder_shape = encoder.output_shape[1:]
    decoder = build_decoder(encoder_shape,filters)

    # 前向
    encoded = encoder(inputs)
    decoded = decoder(encoded)
    outputs = decoded

    return tf.keras.Model(inputs=inputs,outputs=outputs)
auto_encoder = build_autoencoder((2000,1))
auto_encoder.summary()

````

# BI-LSTM Classify

Bidreactional：负责内层复制并输出双向结果,需控制双向数据拼接方式

LSTM:

上一时刻短期记忆与该时刻记忆拼接，输入门、遗忘门、输出门、候选记忆都要从上下文信息中学到，因此需要用可学习的权重控制上述拼接矩阵；否则就像SSM一样直接生成权重矩阵

# LSTM 单个单元：数据流与计算公式

## 1. 单个 LSTM 单元的结构

```text
        当前输入 xₜ          上一步隐藏状态 hₜ₋₁
             │                       │
             └───────────┬───────────┘
                         ↓
                  拼接 zₜ = [hₜ₋₁, xₜ]
                         │
          ┌──────────────┼──────────────┐
          ↓              ↓              ↓
      遗忘门 fₜ       输入门 iₜ       候选记忆 gₜ
          │              │              │
          │              └──────┬───────┘
          │                     ↓
          │                 iₜ ⊙ gₜ
          ↓                     │
    Cₜ₋₁ ⊙ fₜ                  │
          │                     │
          └──────────┬──────────┘
                     ↓
                   相加
                     │
                     ↓
               新细胞状态 Cₜ
                     │
                     ↓ tanh
                     │
                     ⊙ ← 输出门 oₜ
                     │
                     ↓
               新隐藏状态 hₜ
```

注意：四条门控分支是并行计算的，不是当前输入依次通过四个门。

## 2. 四条分支分别计算什么？

令当前输入与上一步隐藏状态拼接为：

\[
z_t=[h_{t-1},x_t]
\]

| 分支 | 计算公式 | 结果是什么 |
|---|---|---|
| 遗忘门 | \(f_t=\sigma(W_fz_t+b_f)\) | 控制旧记忆保留多少 |
| 输入门 | \(i_t=\sigma(W_iz_t+b_i)\) | 控制新内容写入多少 |
| 候选记忆 | \(g_t=\tanh(W_gz_t+b_g)\) | 生成准备写入的新内容 |
| 输出门 | \(o_t=\sigma(W_oz_t+b_o)\) | 控制当前记忆向外输出多少 |

其中：

- \(W_f,W_i,W_g,W_o\) 是模型通过训练学习的参数矩阵。
- \(b_f,b_i,b_g,b_o\) 是可学习的偏置。
- \(\sigma\) 是 Sigmoid 激活函数，输出在 0 和 1 之间。
- \(\tanh\) 是双曲正切激活函数，输出在 -1 和 1 之间。
- \(\odot\) 表示逐元素相乘。


````python
# input_shape = encoder_output_shape
# (N,H,C)
def build_BI_LSTM_Classify(input_shape):
    inputs = L.Input(shape=input_shape)

    temporal_features = L.Bidirectional(
        L.LSTM(
            units=1,
            recurrent_dropout=0.5,
            dropout = 0.5
        ),
        merge_mode = 'concat', # 拼接
    )(inputs)

    x = L.Dense(
        1,
        activation = 'sigmoid'
    )(temporal_features)

    outputs = x

    return tf.keras.Model(inputs=inputs, outputs=outputs)
````

encoder + BI-LSTM Classify

````python
def build_model(input_shape,filters=filters):
    inputs = L.Input(shape=input_shape)

    encoder = build_encoder(input_shape,filters)
    classify = build_BI_LSTM_Classify(encoder.output_shape[1:])

    encoded = encoder(inputs)
    outputs = classify(encoded)

    return tf.keras.Model(inputs,outputs)

m = build_model((2000,1))
m.summary()
````

````python
def predict(model,X):
    prob = model.predict(X).ravel()
    return prob

def firing_power(prob:np.array,n=119):
    prob = (prob >= 0.5).astype(float)
    o = np.concatenate([[0.0], prob]).cumsum()
    idx = np.arange(len(prob))
    return (o[idx+1] - o[np.maximum(0,idx-n+1)]) / n

def alarms(power:np.array,thr=0.5):
    out = (power >= thr).astype(float)
    out = np.flatnonzero(out)
    return out

prob = predict(m,test_f1_data)
power = firing_power(prob)
alarms(power)
````

````python
from sklearn.metrics import (
    roc_auc_score, accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix, balanced_accuracy_score
)

def evaluate(y_true, prob, thr=0.5):
    pred = (prob >= thr).astype(int)

    tn, fp, fn, tp = confusion_matrix(y_true, pred, labels=[0, 1]).ravel()

    sens = tp / (tp + fn) if (tp + fn) else 0.0
    spec = tn / (tn + fp) if (tn + fp) else 0.0
    prec = tp / (tp + fp) if (tp + fp) else 0.0
    f1 = 2 * prec * sens / (prec + sens) if (prec + sens) else 0.0
    acc = (tp + tn) / len(y_true)
    bal_acc = (sens + spec) / 2
    youden = sens + spec - 1

    print("样本数: %d (阳性 %d / 阴性 %d)" % (len(y_true), int((y_true==1).sum()), int((y_true==0).sum())))
    print("AUC       : %.4f" % roc_auc_score(y_true, prob))
    print("Accuracy  : %.4f" % acc)
    print("Balanced  : %.4f" % bal_acc)
    print("Sensitivity: %.4f" % sens)
    print("Specificity: %.4f" % spec)
    print("Precision : %.4f" % prec)
    print("F1        : %.4f" % f1)
    print("Youden J  : %.4f" % youden)
    print("TP=%d FP=%d FN=%d TN=%d" % (tp, fp, fn, tn))

    return dict(auc=roc_auc_score(y_true, prob), acc=acc, bal_acc=bal_acc,
                sens=sens, spec=spec, prec=prec, f1=f1, youden=youden,
                tp=tp, fp=fp, fn=fn, tn=tn)

score = evaluate(y_test,prob,thr=0.5)
for i in score:
    print(f'{i}: {score[i]}')
````
